```python
import pandas as pd

portfolio = pd.DataFrame({

"Company":["TCS","INFY","HDFC"],

"Investment":[100000,90000,80000],

"Current Value":[125000,93000,78000]

})

portfolio
```

```python
portfolio["Profit"] = portfolio["Current Value"] - portfolio["Investment"]

portfolio["Return %"] = (

portfolio["Profit"]

/

portfolio["Investment"]

)*100

portfolio
```

```python
portfolio["Recommendation"] = portfolio["Return %"].apply(

lambda x:

"BUY"

if x>20

else

"HOLD"

if x>5

else

"SELL"

)

portfolio
```
